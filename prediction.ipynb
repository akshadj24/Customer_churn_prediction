```python
import tensorflow as tf
from tensorflow.keras.layers import Dense # type: ignore
from tensorflow.keras.models import  load_model
import pandas as pd
import pickle
import numpy as np
```

```python
#load the ann trained model.standard scalar,enocders

model=load_model('model.h5')

with open('geo_encoder.pkl','rb') as file:
    label_geo=pickle.load(file)

with open('gen_encoder.pkl','rb') as file:
    label_gen=pickle.load(file)    


with open('scalar.pkl','rb') as file:
    scalar=pickle.load(file)
```

```python
input_data={
    "CreditScore":600,
    "Geography":"France",
    "Gender":"Male",
    "Age":40,
    "Tenure":3,
    "Balance":60000,
    "NumOfProducts":2,
    "HasCrCard":1,
    "IsActiveMember":1,
    "EstimatedSalary":50000
}
```

```python
df_geo=label_geo.transform([[input_data['Geography']]]).toarray()
l1=pd.DataFrame(df_geo,columns=label_geo.get_feature_names_out(['Geography']))
```

```python
l1.first
```

```python
[[input_data['Gender']]]
```

```python
df_gen=label_gen.transform([[input_data['Gender']]])
l2=pd.DataFrame(df_gen,columns=["Gender"])
```

```python
input_data['Gender']=int(l2['Gender'])
```

```python
input_data['Geography']=l1.iloc[0].values.tolist()
```

```python
input_data['Geography']="France"
```

```python
input_data
```

```python
input_df=pd.DataFrame([input_data])
```

```python
input_df=pd.concat([input_df,l1],axis=1)
input_df.head()
```

```python
input_df['Gender']=label_gen.transform(input_df['Gender'])
input_df
```

```python
input_df.drop("Geography",axis=1,inplace=True)
```

```python
input_scaled=scalar.transform(input_df)
input_scaled
```

```python
prediction=model.predict(input_scaled)
```
